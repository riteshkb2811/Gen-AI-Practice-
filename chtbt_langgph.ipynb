```python
from dotenv import load_dotenv
from pydantic import BaseModel
from langchain_groq import ChatGroq
from langgraph.graph import StateGraph,START,END
from langgraph.checkpoint.memory import InMemorySaver
from typing import Annotated
```

```python
from langgraph.graph import add_messages
import os
class ChatState(BaseModel):
    messages:Annotated[list,add_messages]
load_dotenv()
key1=os.getenv("GroqKEY")
llm1=ChatGroq(model="openai/gpt-oss-20b",groq_api_key=key1)

def chatBotNode(state:ChatState)-> ChatState:
    res = llm1.invoke(state.messages)
    return {"messages": [res]}

graph=StateGraph(ChatState)
graph.add_node("Chatbot",chatBotNode)

graph.add_edge(START,"Chatbot")
graph.add_edge("Chatbot",END)
inmem=InMemorySaver()
app=graph.compile(checkpointer=inmem)

```

```python
while True:
 query=input("User: ")
 if query.lower() =="exit":
    break
 res = app.invoke(
    {"messages": [{"role": "user", "content":query}]},
    {"configurable": {"thread_id": "Ritesh"}}
 )
 print("")
 print("AI: ",res["messages"][-1].content)
 print("")
```
