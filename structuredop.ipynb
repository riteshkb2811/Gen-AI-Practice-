```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
app1=os.getenv("Gemini_KEY")
llm= ChatGoogleGenerativeAI(model="gemini-3.6-flash",google_api_key=app1)
```

```python
text="Hello,my name is ritesh and email is xyz@gmail.com and location is pune"
res=llm.invoke(f"Give me th name examil and location in text form from {text}")
print(res.text)
```

```python
from pydantic import BaseModel,Field
from typing import List 
class Responsemodel(BaseModel):
    name:str=Field(description="Name of the User")
    email:str=Field(description="Email Address")
    location:str=Field(description="Location")


struct_llm=llm.with_structured_output(Responsemodel)
res=struct_llm.invoke(f"Give me th name examil and location in text form from {text}").model_dump_json
```

```python
class movies(BaseModel):
    name:str
    year:int

class allmovies(BaseModel):
    allmovies:List[movies]

struct=llm.with_structured_output(movies)
res1=struct.invoke("Give the latest two movies in 2026 in marathi languague").model_dump_json
print(res1)
```

```python
from langchain_core.tools import tool
@tool
def add(a:int,b:int):
    """
    it will return addition of two numbers
    """
    return a+b

@tool
def mul(a:int,b:int):
    """
    it will return multiplication of two numbers
    """
    return a*b
```

```python
print(add.invoke({"a":10,"b":20}))
```

```python
from pyexpat import model
from langchain.agents import create_agent
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
groq_key = os.getenv("GroqKEY")
llm = ChatGroq(model="openai/gpt-oss-20b", groq_api_key=groq_key,streaming=True)

agent1=create_agent(
    model=llm,
    tools=[add,mul],
    system_prompt="You are a math professor and use tools for calculation"
)

respo=agent1.invoke({"messages":[{"role":"user","content":"what is addition of 1345 and 3456 ?"}]})

# for chunk in respo['messages'][-1]:
#     print(chunk)
#     print(end="")

print(respo["messages"][-1].content)
```

```python
from langchain_tavily import TavilySearch
from langchain.agents import create_agent
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
groq_key = os.getenv("GroqKEY")
tavilly_key=os.getenv("TAVILY_API_KEY")
llm = ChatGroq(model="openai/gpt-oss-20b", groq_api_key=groq_key,streaming=True)

tavily_search_tool = TavilySearch(
    max_results=5,
    topic="general",
    tavilly_api_key=tavilly_key
)

agent = create_agent(model=llm,
                  tools=[tavily_search_tool],
                  system_prompt="you are searcher and search any question on Tavilly")


res = agent.invoke({"messages": [("user", "what is the latest price of gold in indian rupees in marathi")]})


for chunk in res["messages"]:
    print(chunk)
    print(end="")

print(res["messages"][-1].content)
```
