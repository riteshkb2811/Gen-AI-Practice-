```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

# Load your .env file
load_dotenv()

# Initialize Gemini
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=os.getenv("Gemini_KEY")  # matches Gemini_KEY in your .env
)

# Test invocation
response = llm.invoke("Hi! Introduce yourself in one sentence.")
print(response.content)


prompts = [
    {"role": "system", "content": "You are an airline pilot."},
    {"role": "user", "content": "ATC: Avoid landing due to severe crosswinds. Confirm and report your action."}
]


ans=llm.invoke(prompts)
print(ans.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser 

out=StrOutputParser()


prompt_template = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are an ATC"},
    {"role": "user", "content": "Pilot: {Query}"}
])


formatted_messages = prompt_template.format_messages(
    Query="hi this is romeo india tango echo seirra hotel ,taxi complete, on standby, requesting clearnce for rtake off"
)
ans = llm.invoke(formatted_messages)
print(ans.content)


def lowercse(query:str):
    return query.lower()
```

```python
chain = prompt_template | llm | out | lowercse

res = chain.invoke({"Query": "on 32 lima and runway 27"})
print(res)
```

```python
history =[]
while True:
    query=input("User : ")
    if query.lower in ["exit","quit","bye"]:
        print("Good Bye")
        break
    history.append({"role":"user","content":query})
    print("User: ",query)

    res=llm.invoke(history)
    history.append({"role":"ai","content":query})
    print("AI: ",res.text,"\n")
    
```
