```python
from pydantic import BaseModel,Field
from typing import List 
```

```python
class abc(BaseModel):
    name:str
    age:int
    gender:str
```

```python
abc.name="Ritesh"
abc.age=23
abc.gender="Male"
```

```python
print(abc.name)
print(abc.age)
print(abc.gender)
```

```python
class abc(BaseModel):
    name:str=Field(description="name")
    age:int=Field(description="Age")
    gender:str=Field(description="Gender")
```

```python
class clg(BaseModel):
    clg:str
    student:abc


```
